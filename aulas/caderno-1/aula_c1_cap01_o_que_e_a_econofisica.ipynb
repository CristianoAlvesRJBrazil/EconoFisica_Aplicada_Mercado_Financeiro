## 1. Uma definição

A **Econofísica** é a área que aplica métodos da física, sobretudo da física estatística e
da teoria dos sistemas complexos, a problemas da economia e das finanças. O termo se
consolidou na segunda metade dos anos 1990 e é costumeiramente atribuído a H. Eugene
Stanley. Um dos primeiros livros a organizar a área foi o de Mantegna e Stanley (1999).

A definição diz menos sobre os temas, que a economia e as finanças estudam há muito tempo,
e mais sobre a **forma** de estudá-los:

| Sistema físico | Mercado financeiro |
|---|---|
| Um número enorme de partículas | Um número enorme de participantes |
| As partículas interagem | As decisões interagem |
| Comportamento coletivo: pressão, magnetização | Resultado coletivo: o preço |
| Estudado sem acompanhar cada partícula | Estudado sem acompanhar cada investidor |

A pergunta da Econofísica é quanto da estatística dos preços pode ser entendida com as
ideias que funcionaram para esses sistemas físicos.

## 2. Por que físicos olharam para os mercados

Três características dos mercados financeiros os tornaram atraentes para a física.

1. **Dados abundantes.** Poucos sistemas complexos são tão bem medidos quanto um mercado.
   Cada negócio tem preço, quantidade e horário registrados, e séries diárias de índices
   cobrem décadas. Com tantos dados, é possível medir propriedades estatísticas finas,
   como a forma das caudas de uma distribuição, e confrontá-las com modelos.
2. **Fenômenos coletivos.** Os mercados exibem comportamentos que não se explicam olhando um
   participante isolado: pânicos, bolhas, períodos de calma e de agitação. Na física,
   fenômenos assim são estudados com modelos de muitos agentes em interação, e é
   exatamente esse tipo de modelo que o Módulo 8 constrói.
3. **Leis de escala.** A física dos fenômenos críticos mostrou que sistemas muito diferentes
   podem obedecer às mesmas leis de potência, com os mesmos expoentes. Encontrar leis de
   escala nos dados financeiros, e perguntar se elas são **universais**, isto é, as mesmas
   para mercados e ativos diferentes, é um dos programas centrais da área.

Depois de baixar os dados (Seção 5), a Seção 6 mostra as duas primeiras características
nos números.

## 3. Objetos de estudo e métodos

Os métodos da Econofísica que este curso cobre, a pergunta que cada um ajuda a responder e
o módulo em que aparecem:

| Método | Pergunta | Módulo |
|---|---|:-:|
| Distribuições de cauda pesada | Com que frequência acontecem os retornos extremos? | 2 |
| Processos estocásticos | Que modelo de dinâmica gera os preços observados? | 3 |
| Leis de potência e escala | As caudas e as flutuações seguem leis de escala? | 4 |
| Fractais e multifractais | Os retornos têm memória longa? Como ela escala no tempo? | 5 |
| Matrizes aleatórias e redes | Quanto das correlações entre ativos é sinal, e quanto é ruído? | 6 |
| Entropia e informação | Quão previsível é uma série? Quem influencia quem? | 7 |
| Modelos de agentes | Que interações entre investidores geram os fatos observados? | 8 |
| Pontos críticos e extremos | Como medir o risco de crises? Bolhas têm assinatura? | 9 |

Três atitudes atravessam todos esses métodos:

- **Empirismo:** o ponto de partida são as regularidades medidas nos dados, e não hipóteses
  sobre o comportamento dos investidores.
- **Modelos mínimos:** o modelo mais simples que reproduz um fenômeno ensina mais do que um
  modelo com muitos parâmetros ajustados.
- **Simulação:** quando um modelo não tem solução analítica, ele é explorado no computador,
  como você fez com o passeio aleatório no Caderno 0.

## 4. Econofísica, economia e finanças

A Econofísica **não substitui** a economia nem as finanças:

| Área | Pergunta central |
|---|---|
| Teoria econômica | *Por que* os agentes agem como agem, e qual é o equilíbrio resultante? |
| Finanças | Como apreçar ativos e montar carteiras? |
| Econometria financeira | Que modelos estatísticos rigorosos descrevem as séries de preços? |
| Econofísica | *Que leis estatísticas* os preços obedecem, e que mecanismos coletivos podem produzi-las? |

As abordagens frequentemente estudam os mesmos fenômenos. O agrupamento de volatilidade,
por exemplo, foi medido por econofísicos com expoentes de escala e por econometristas com
modelos da família GARCH (Capítulo 5). Este curso trata a Econofísica como complemento:
sempre que possível, os resultados serão comparados com os da abordagem clássica.

> **Contexto de mercado.** O resto do caderno ilustra essa relação. O Capítulo 2 mostra que
> a ideia de passeio aleatório veio de um matemático e de um físico; o Capítulo 3, que ela
> virou o centro da teoria financeira; e o Capítulo 4, onde ela falha, no que os dados
> mostram.

### 5.1 O módulo de dados

Todos os capítulos precisam dos mesmos log-retornos. Por isso, as funções que leem os
arquivos ficam num módulo, `dados_caderno1.py`, na pasta de código do caderno:

```python
"""Funções de dados usadas em todos os capítulos do Caderno 1."""
from pathlib import Path

import numpy as np
import pandas as pd


def arquivo_mais_recente(padrao):
    """Devolve a cópia mais recente de um arquivo (nomes com data ISO)."""
    copias = sorted(Path("dados").glob(padrao))
    if not copias:
        raise FileNotFoundError(f"Sem arquivo {padrao}: rode a Listagem 1.1")
    return copias[-1]


def log_retornos(precos):
    """Log-retornos de uma série de preços, só nos dias com cotação."""
    return np.log(precos.dropna()).diff().dropna()


def carregar_retornos():
    """Log-retornos diários do S&P 500 e do Ibovespa."""
    sp500 = pd.read_csv(arquivo_mais_recente("yahoo_sp500_*.csv"),
                        index_col="Date", parse_dates=True)["^GSPC"]
    ibov = pd.read_csv(arquivo_mais_recente("yahoo_ibovespa_*.csv"),
                       index_col="Date", parse_dates=True)["^BVSP"]
    return log_retornos(sp500), log_retornos(ibov)
```

## 1. Variáveis, tipos e operações

Uma **variável** é um nome associado a um valor. O sinal `=` faz a associação, e o Python
deduz o **tipo** do valor: `int` para números inteiros, `float` para números com casas
decimais, `str` para textos e `bool` para verdadeiro ou falso.

**Listagem 2.1.** Uma posição em ações: preço, quantidade e valor.

````python
# Uma posição em ações
preco = 36.77         # R$ por ação: fechamento da PETR4 em 02/01/2025
quantidade = 100      # número de ações
valor = preco * quantidade

print(valor)
print(type(preco), type(quantidade), type(valor))
````

<class 'float'> <class 'int'> <class 'float'>

O resultado da primeira linha surpreende: `3677.0000000000005`, e não `3677`. Não é um

> **Armadilha comum.** Números `float` são guardados em base 2, com cerca de 16 dígitos
> significativos. Valores como 0,1 ou 36,77 não têm representação exata em base 2, e
> pequenos erros de arredondamento aparecem nas contas. Por isso, **nunca compare dois
> `float` com `==`**: use `math.isclose`, que admite uma pequena tolerância. Para exibir
> valores, arredonde com `round` ou na formatação.

## 4. Decisões e repetições

O laço `for` repete um bloco de código para cada elemento de uma sequência. O bloco é
definido pela **indentação**: as linhas recuadas em quatro espaços. A listagem abaixo
calcula o **retorno simples** de cada pregão,

$$r_t = \frac{P_t}{P_{t-1}} - 1,$$

em que $P_t$ é o preço de fechamento no pregão $t$.

**Listagem 2.6.** Retorno simples de cada pregão com um laço `for`.

````python
# Retorno simples de cada pregão: r = P(t) / P(t-1) - 1
retornos = []
for i in range(1, len(fechamentos)):
    r = fechamentos[i] / fechamentos[i - 1] - 1
    retornos.append(r)

for r in retornos:
    print(f"{r:+.2%}")
````

**Saída esperada:**

```text
-1.06%
-0.47%
+2.13%
-0.81%
+0.44%
```

As decisões usam `if`, `elif` ("senão, se") e `else`. O laço `while` repete enquanto uma
condição for verdadeira, e é útil quando não se sabe de antemão quantas repetições serão
necessárias.

**Listagem 2.7.** Contando altas e quedas; um laço `while`.

````python
altas = quedas = 0
for r in retornos:
    if r > 0:
        altas += 1
    elif r < 0:
        quedas += 1
print(f"{altas} altas e {quedas} quedas")

# while: em que pregão o preço voltou ao primeiro fechamento?
i = 1
while fechamentos[i] < fechamentos[0]:
    i += 1
print(f"Voltou a R$ {fechamentos[0]} ou mais no pregão de índice {i}")
````

**Saída esperada:**

```text
2 altas e 3 quedas
Voltou a R$ 36.77 ou mais no pregão de índice 3
```

> **Armadilha comum.** Um `while` cuja condição nunca fica falsa repete para sempre. Aqui,
> se o preço nunca voltasse ao primeiro fechamento, o índice `i` passaria do fim da lista e
> o programa terminaria com um erro. Sempre pergunte: o que garante que este laço termina?

*Experimente:* na célula acima, troque `fechamentos[0]` por `40` no `while` e execute. O
erro `IndexError` mostra o índice passando do fim da lista. Depois, desfaça a troca e
execute a célula de novo.

Muitas vezes um laço só serve para montar uma lista nova. A **compreensão de lista** faz
isso numa expressão só, e ainda permite filtrar elementos com `if`.

**Listagem 2.8.** Compreensões de lista: retornos e só as quedas.

````python
# Compreensão de lista: o mesmo cálculo da Listagem 2.6 em uma expressão
retornos = [fechamentos[i] / fechamentos[i - 1] - 1
            for i in range(1, len(fechamentos))]
quedas = [r for r in retornos if r < 0]
print(len(retornos), len(quedas))
print([round(r, 4) for r in retornos])
````

**Saída esperada:**

```text
5 3
[-0.0106, -0.0047, 0.0213, -0.0081, 0.0044]
```

## 5. Funções

Uma **função** dá nome a um cálculo, para que ele seja escrito uma vez e usado muitas. A
definição começa com `def`; as **anotações de tipo** (`p0: float`, `-> float`) dizem o que
a função recebe e devolve; e a **docstring**, entre aspas triplas, explica o que ela faz.

O **retorno acumulado** de um período compõe os retornos de cada pregão:

$$R = (1 + r_1)(1 + r_2)\cdots(1 + r_n) - 1 = \frac{P_n}{P_0} - 1.$$

A listagem implementa as duas formas e usa `assert` para conferir que elas coincidem.

**Listagem 2.9.** Funções para o retorno simples e o retorno acumulado.

````python
def retorno_simples(p0: float, p1: float) -> float:
    """Retorno simples entre dois preços: p1 / p0 - 1."""
    return p1 / p0 - 1


def retorno_acumulado(retornos: list[float]) -> float:
    """Compõe retornos simples: (1 + r1)(1 + r2)...(1 + rn) - 1."""
    fator = 1.0
    for r in retornos:
        fator *= 1 + r
    return fator - 1


print(f"{retorno_simples(36.77, 36.84):.4%}")
print(f"{retorno_acumulado(retornos):.4%}")
assert math.isclose(retorno_acumulado(retornos),
                    retorno_simples(fechamentos[0], fechamentos[-1]))
print("Conferido: compor os retornos diários dá o retorno do período.")
````

**Saída esperada:**

```text
0.1904%
0.1904%
Conferido: compor os retornos diários dá o retorno do período.
```

> **Boa prática.** Teste as suas funções com casos cujo resultado você conhece. O `assert`
> interrompe o programa se a condição for falsa: se uma alteração futura quebrar a função,
> você descobre na hora, e não depois de publicar um resultado errado.

### 5.1 A docstring é a ajuda da função

`help` mostra a assinatura, com as anotações de tipo, e a docstring:

````python
help(retorno_acumulado)
````

Um `assert` falso para o programa com um `AssertionError` e a mensagem escrita depois da
vírgula. Mais uma vez, o `try`/`except` só evita que o notebook pare:

````python
try:
    assert retorno_simples(36.77, 36.84) > 0.01, "esperava uma alta de mais de 1%"
except AssertionError as erro:
    print("AssertionError:", erro)
````

**Saída esperada:**

```text

```

## 6. A biblioteca padrão: matemática, datas e arquivos

O Python vem com uma **biblioteca padrão** extensa. Três dos seus módulos aparecem o tempo
todo em finanças.

**`math`.** Funções matemáticas, como o logaritmo natural. O **retorno logarítmico** entre
dois preços é

$$\ell = \ln\frac{P_1}{P_0} = \ln(1 + r).$$

Para variações pequenas, $\ln(1 + r) \approx r$, e os dois retornos quase coincidem; para
variações grandes, não. O Capítulo 6 mostra por que o retorno logarítmico é o preferido na
Econofísica.

**Listagem 2.10.** Retorno simples e retorno logarítmico.

````python
# Retorno logarítmico: ln(p1 / p0). Para variações pequenas, fica muito
# perto do retorno simples; para variações grandes, não.
for p1 in [36.84, 40.00, 50.00]:
    simples = retorno_simples(36.77, p1)
    logaritmico = math.log(p1 / 36.77)
    print(f"R$ {p1:5.2f}   simples {simples:+.4f}   log {logaritmico:+.4f}")
````

**Saída esperada:**

```text

R$ 40.00   simples +0.0878   log +0.0842

```

## 7. Erros e exceções

Quando algo dá errado, o Python interrompe o programa com uma **exceção**, que tem um tipo
e uma mensagem. Um caso clássico com dados brasileiros: planilhas e arquivos exportados em
português usam vírgula decimal, e o Python não converte esse texto em número.

**Listagem 2.13.** Um erro de propósito: converter um número com vírgula decimal.

```python
float("36,77")      # vírgula decimal, como nas planilhas brasileiras
```

Executada sozinha, essa linha interrompe o notebook. Para que *Run All Cells* não pare
aqui, a célula abaixo executa a mesma linha dentro de um `try`/`except` e imprime só a
última linha da mensagem de erro, como no caderno.

````python
try:
    float("36,77")      # vírgula decimal, como nas planilhas brasileiras
except ValueError as erro:
    print(f"{type(erro).__name__}: {erro}")
````

**Saída esperada:**

```text
ValueError: could not convert string to float: '36,77'
```

*Experimente:* para ver a mensagem completa (o *traceback*), tire o `#` do começo da linha
abaixo e execute a célula. Leia a mensagem de baixo para cima: a última linha diz o tipo
do erro e o motivo. Depois, ponha o `#` de volta.

````python
# float("36,77")
````

O bloco `try`/`except` captura a exceção e decide o que fazer. A função abaixo troca a
vírgula por ponto e, se mesmo assim não conseguir converter, devolve `None`, o valor que o
Python usa para "nada".

**Listagem 2.14.** Tratando a exceção: conversão tolerante a vírgulas.

````python
def para_numero(texto: str) -> float | None:
    """Converte '36,77' ou '36.77' em número; devolve None se não der."""
    try:
        return float(texto.replace(",", "."))
    except ValueError:
        return None


for texto in ["36,77", "36.38", "n/d", ""]:
    print(repr(texto), "->", para_numero(texto))
````

**Saída esperada:**

```text
'36,77' -> 36.77
'36.38' -> 36.38
'n/d' -> None
'' -> None
```

> **Armadilha comum.** Capturar exceções demais. Um `except:` sem tipo esconde qualquer
> erro, inclusive os que indicam um defeito no seu código. Capture só o tipo que você
> espera (`ValueError`, aqui) e deixe os outros aparecerem.

## 8. Prática guiada: a primeira semana de 2025

O código abaixo junta tudo o que a aula mostrou para montar um relatório de três ações:

- **dicionário** de listas com os fechamentos (Seção 3);
- **compreensões de lista** para as datas e os retornos (Seção 4);
- as **funções** `retorno_simples` e `retorno_acumulado` (Seção 5);
- **datas** no padrão brasileiro (Seção 6);
- **f-strings** com largura fixa para alinhar a tabela (Seção 2).

Leia a função antes de executar e tente prever a saída da PETR4: você já conhece os
retornos dela.

````python
from datetime import date

datas = ["2025-01-02", "2025-01-03", "2025-01-06",
         "2025-01-07", "2025-01-08", "2025-01-09"]
semana = {                                  # fechamentos (fonte: B3)
    "PETR4": [36.77, 36.38, 36.21, 36.98, 36.68, 36.84],
    "VALE3": [54.25, 53.24, 52.56, 52.05, 51.55, 51.23],
    "ITUB4": [30.57, 29.82, 31.16, 31.50, 30.99, 31.07],
}


def relatorio(ticker: str, datas: list[str], precos: list[float]) -> None:
    """Imprime os fechamentos, o retorno de cada pregão e o do período."""
    dias = [date.fromisoformat(d).strftime("%d/%m/%Y") for d in datas]
    rets = [retorno_simples(precos[i - 1], precos[i])
            for i in range(1, len(precos))]
    print(f"{ticker:<12}{'Fechamento':>11}{'Retorno':>10}")
    print(f"{dias[0]:<12}{precos[0]:>11.2f}")
    for i in range(1, len(precos)):
        print(f"{dias[i]:<12}{precos[i]:>11.2f}{rets[i - 1]:>+10.2%}")
    altas = len([r for r in rets if r > 0])
    print(f"Período: {retorno_acumulado(rets):+.2%}, "
          f"com {altas} altas em {len(rets)} pregões")


for ticker, serie in semana.items():
    relatorio(ticker, datas, serie)
    print()
````

Execute primeiro a célula a seguir, que define a conferência.

````python
def conferir(nome, teste):
    """Roda o teste de um exercício e diz se passou, sem interromper o notebook."""
    try:
        teste()
    except NotImplementedError:
        print(f"{nome}: ainda não resolvido")
    except AssertionError as erro:
        print(f"{nome}: confira. {erro}")
    except Exception as erro:
        print(f"{nome}: a solução deu erro. {type(erro).__name__}: {erro}")
    else:
        print(f"{nome}: ok")


def igual(obtido, esperado):
    """Compara números com tolerância; outros valores, com ==."""
    if isinstance(esperado, float) and isinstance(obtido, (int, float)):
        return math.isclose(obtido, esperado, abs_tol=1e-9)
    return obtido == esperado
````
